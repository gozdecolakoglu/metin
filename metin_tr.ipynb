# **Metni küçük harfe çevir**

```python
metin = metin.lower()
print(metin)
```

# **Noktalama işaretlerini temizle**

```python
import re
metin = re.sub(r'[^\w\s]', '', metin)
print(metin)
```

## **Tokenization yap**
Metni kelimelere ayır

```python
import nltk
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

kelimeler = word_tokenize(metin, language="turkish")

print(kelimeler)
print("Toplam kelime sayısı:", len(kelimeler))
```

# **Stop-word'leri temizle**

```python
stop_words = set([
    "ve", "veya", "ile", "bir", "bu", "şu", "o",
    "da", "de", "için", "ama", "fakat", "çok",
    "daha", "gibi", "olan"
])

temiz_kelimeler = [
    kelime for kelime in kelimeler
    if kelime not in stop_words
]

print("Stop-word temizleme sonrası:")
print(temiz_kelimeler)

print("\nKelime sayısı:", len(temiz_kelimeler))
```

# **Turkish Stemmer ile kök bul**

```python
from TurkishStemmer import TurkishStemmer

stemmer = TurkishStemmer()

kokler = [stemmer.stem(kelime) for kelime in temiz_kelimeler]

print(kokler)
```
