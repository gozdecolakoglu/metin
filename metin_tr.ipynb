# **Noktalama işaretlerini temizle**

```python
import re
metin = re.sub(r'[^\w\s]', '', metin)
print(metin)
```

## **Tokenization yap**
Metni kelimelere ayır

```python
import nltk
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

kelimeler = word_tokenize(metin, language="turkish")

print(kelimeler)
print("Toplam kelime sayısı:", len(kelimeler))
```

# **Stop-word'leri temizle**

```python
stop_words = set([
    "ve", "veya", "ile", "bir", "bu", "şu", "o",
    "da", "de", "için", "ama", "fakat", "çok",
    "daha", "gibi", "olan"
])

temiz_kelimeler = [
    kelime for kelime in kelimeler
    if kelime not in stop_words
]

print("Stop-word temizleme sonrası:")
print(temiz_kelimeler)

print("\nKelime sayısı:", len(temiz_kelimeler))
```

# **Turkish Stemmer ile kök bul**

```python
from TurkishStemmer import TurkishStemmer

stemmer = TurkishStemmer()

kokler = [stemmer.stem(kelime) for kelime in temiz_kelimeler]

print(kokler)
```

```python
# 8. TURKISHSTEMMER İLE STEMMING
# ============================================================

stemmer = TurkishStemmer()

stemler = [
    stemmer.stem(kelime)
    for kelime in temiz_kelimeler
]

print("\nİlk 20 stemming sonucu:")
for kelime, stem in zip(temiz_kelimeler[:20], stemler[:20]):
    print(f"{kelime:25} -> {stem}")
```

# **TÜM TEMİZ KELİMELER İÇİN LEMMATIZATION**

```python
kokler = []
lemmalar = []
analizler = []

for kelime in temiz_kelimeler:
    try:
        # Kelimeyi Zemberek ile analiz et
        analiz = morphology.analyze(kelime)
        results = analiz.getAnalysisResults()

        # Sonuç listesi boş değilse ilk analizi al
        if results.size() > 0:
            sonuc = results.get(0)

            # ------------------------------------------------
            # KÖK (getStems() listesinden ilk kökü al)
            # ------------------------------------------------
            stems = sonuc.getStems()
            kok = str(stems.get(0)) if stems.size() > 0 else kelime

            # ------------------------------------------------
            # LEMMA
            # getLemmas() listesinin son veya ilk elemanı
            # ya da sözlük kök adı: sonuc.getDictionaryItem().lemma
            # ------------------------------------------------
            lemmas = sonuc.getLemmas()
            if lemmas.size() > 0:
                # getLemmas() genelde çekim/türetim aşamalarını verir,
                # temel sözlük biçimi için ilk eleman veya getDictionaryItem() kullanılır:
                lemma = str(sonuc.getDictionaryItem().lemma)
            else:
                lemma = kelime

            # ------------------------------------------------
            # MORFOLOJİK ANALİZ
            # ------------------------------------------------
            analiz_bilgisi = str(sonuc.formatLong())

        else:
            kok = kelime
            lemma = kelime
            analiz_bilgisi = "Analiz bulunamadı"

    except Exception as e:
        kok = kelime
        lemma = kelime
        analiz_bilgisi = f"Analiz hatası: {e}"

    kokler.append(kok)
    lemmalar.append(lemma)
    analizler.append(analiz_bilgisi)


# ============================================================
# SONUÇLARI GÖSTER
# ============================================================

print("=" * 80)
print("ZEMBEREK KÖK VE LEMMATIZATION SONUÇLARI")
print("=" * 80)

for kelime, kok, lemma in zip(temiz_kelimeler, kokler, lemmalar):
    print(f"{kelime:25} | Kök: {kok:20} | Lemma: {lemma}")

print("=" * 80)
print("Toplam kelime:", len(temiz_kelimeler))
```

```python
import pandas as pd

# Lemmatization sonuçlarını bir DataFrame'e dönüştür
results_df = pd.DataFrame({
    'Orijinal Kelime': temiz_kelimeler,
    'Kök': kokler,
    'Lemma': lemmalar
})

# DataFrame'i göster
display(results_df.head())
```

```python
# DataFrame'i CSV dosyasına kaydet
output_filename = 'zemberek_lemmatization_results.csv'
results_df.to_csv(output_filename, index=False, encoding='utf-8')

print(f"Sonuçlar '{output_filename}' adlı dosyaya kaydedildi.")
```
