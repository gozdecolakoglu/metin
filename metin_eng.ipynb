```python
metin="Artificial intelligence has become one of the fastest-developing fields of technology in recent years. Today, artificial intelligence applications are used not only in technology companies but also in many different sectors, from education to healthcare, from finance to transportation. Many of the processes that people perform in their daily lives are supported by artificial intelligence systems without them being aware of it. For example, voice assistants used on smartphones, recommendation systems on the internet, and personalized products offered by online shopping platforms are some of the areas where artificial intelligence is used in daily life.One of the important areas of application of artificial intelligence is the education sector. Systems that analyze students' learning processes can recommend different content according to students' needs. Teachers, on the other hand, can examine students' performance and determine which subjects they need more support in. Especially on online education platforms, intelligent systems can analyze students' study habits and create personalized learning experiences. This contributes to making the education process more efficient and student-centered.Artificial intelligence technologies are also used in the healthcare sector. By analyzing large amounts of data collected in hospitals, it can be possible to help detect diseases at an early stage. Artificial intelligence-supported applications are used in processes such as examining medical images, evaluating patient records, and identifying risky situations. However, it is of great importance that the systems used in the healthcare field are reliable and that personal data is protected.From a business perspective, artificial intelligence is an important tool that supports decision-making processes. Companies try to predict future demand by examining their customers' purchasing behaviors. Data analyses conducted on issues such as customer churn, sales volume, product preferences, and customer satisfaction can help businesses make more accurate decisions. In addition, chatbots can quickly answer customers' questions, enabling customer services to be carried out more effectively.However, the widespread use of artificial intelligence also brings some important problems. Personal data security, algorithmic biases, the generation of false information, and the transformation of some professions are among the issues discussed in society. Therefore, when developing artificial intelligence systems, attention should be paid not only to technical success but also to ethical and legal responsibilities.It is thought that artificial intelligence will have a greater place in people's lives in the future. With the development of new technologies, the interaction between people and artificial intelligence systems will also change. During this process, it is important for individuals to improve their digital literacy skills, evaluate the opportunities offered by technology correctly, and be aware of potential risks. In order for artificial intelligence to be used as a tool that makes human life easier, the balance between technology, people, and society needs to be maintained."
print(metin)
```

# **Gerekli Kütüphaneler**

```python
import string
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer

# NLTK bileşenlerini indirme
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

# İngilizce Zemberek karşılığı: spaCy (Morfolojik Analiz & Lemmatization)
try:
    import spacy
    nlp = spacy.load("en_core_web_sm")
    spacy_available = True
except Exception:
    spacy_available = False
```

# **4. METİN ÖN İŞLEME (PREPROCESSING)**

# **4.1 Küçük harfe dönüştürme (Lowercasing)**

```python
metin_lower = metin.lower()
print(metin_lower)
```

# **4.2 Noktalama temizleme (Punctuation Removal)**

```python
metin_no_punct = metin_lower.translate(str.maketrans('', '', string.punctuation))
print(metin_no_punct)
```

# **4.3 Tokenization (Kelimelere ayırma)**

```python
tokens = word_tokenize(metin_no_punct)
print(tokens)
```

# **4.4 Stop-word temizleme (Gereksiz kelimeleri filtreleme)**

```python
stop_words = set(stopwords.words('english'))
clean_tokens = [w for w in tokens if w not in stop_words and len(w) > 1]

print(f"4. Ön İşleme İstatistikleri:")
print(f" - Toplam Token Sayısı          : {len(tokens)}")
print(f" - Temizlenmiş Kelime Sayısı     : {len(clean_tokens)}")
print(f" - İlk 8 Temiz Kelime Örneği     : {clean_tokens[:8]}")
print("-" * 70)
```

# **5.İngilizce Stemmer ile kök bulma(Porter vs Snowball Stemmer)**

```python
porter = PorterStemmer()
snowball = SnowballStemmer('english')

print("5. İngilizce Stemmer (Porter vs Snowball) - İlk 8 Kelime:")
print(f"{'Orijinal Kelime':<18} {'Porter Stemmer':<18} {'Snowball Stemmer':<18}")
print("-" * 55)
for word in clean_tokens[:8]:
    print(f"{word:<18} {porter.stem(word):<18} {snowball.stem(word):<18}")
print("-" * 70)
```

# **6.Morfolojik Analiz (spaCy)**

```python
print("6. İngilizce Morfolojik Analiz & Kök Çözümleme (Zemberek Eşdeğeri: spaCy):")

if spacy_available:
    sample_sentence = "Artificial intelligence applications are transforming modern healthcare systems."
    doc = nlp(sample_sentence)

    print(f"Örnek Cümle Analizi: \"{sample_sentence}\"\n")
    print(f"{'Kelime':<15} {'Kök (Lemma)':<15} {'Kelime Türü (POS)':<20} {'Morfolojik Detay (Morph)'}")
    print("-" * 75)
    for token in doc:
        if not token.is_punct:
            print(f"{token.text:<15} {token.lemma_:<15} {token.pos_:<20} {str(token.morph)}")
else:
    print(" [!] spaCy yüklü değil. Kurmak için: pip install spacy && python -m spacy download en_core_web_sm")
print("-" * 70)
```

# **7. Tek Kelimelik Test**

```python
test_word = "developing"
wn_lemmatizer = WordNetLemmatizer()

print(f"7. Tek Kelimelik Karşılaştırma Testi: '{test_word}'")
print(f" - Porter Stemmer             : {porter.stem(test_word)}")
print(f" - Snowball Stemmer           : {snowball.stem(test_word)}")
print(f" - NLTK WordNet (Varsayılan)  : {wn_lemmatizer.lemmatize(test_word)}")
print(f" - NLTK WordNet (Fiil olarak) : {wn_lemmatizer.lemmatize(test_word, pos='v')}")
if spacy_available:
    spacy_test = nlp(test_word)[0]
    print(f" - spaCy Morfolojik Kök       : {spacy_test.lemma_} (Tür: {spacy_test.pos_})")
print("-" * 70)
```

# **8. TÜM TEMİZ KELİMELER İÇİN LEMMATIZATION**

```python
lemmatized_tokens = [wn_lemmatizer.lemmatize(w) for w in clean_tokens]

print("8. Tüm Temiz Kelimeler İçin Lemmatization (İlk 12 Kelime):")
print(f"{'Temiz Kelime':<20} {'Lemmatized (Kök)':<20}")
print("-" * 40)
for orig, lemma in zip(clean_tokens[:12], lemmatized_tokens[:12]):
    print(f"{orig:<20} {lemma:<20}")

print("-" * 70)
print("Tüm işlemler başarıyla tamamlandı.")
```

```python
# tüm kelimeler için
lemmatized_tokens = [wn_lemmatizer.lemmatize(w) for w in clean_tokens]

print(f"8. Tüm Temiz Kelimeler İçin Lemmatization (Toplam {len(clean_tokens)} Kelime):")
print(f"{'Sıra':<6} {'Temiz Kelime':<24} {'Lemmatized (Kök)':<24}")
print("-" * 55)

for i, (orig, lemma) in enumerate(zip(clean_tokens, lemmatized_tokens), start=1):
    print(f"{i:<6} {orig:<24} {lemma:<24}")

print("-" * 55)
print("Tüm kelimeler başarıyla işlendi ve listelendi.")
```
